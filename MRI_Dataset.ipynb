```python
import gdown
import os
import subprocess

file_id = "12gxEbo-354fww46f88Sl_Nq2x9cU7eiH" # Assuming this is the correct file ID for the RAR file
rar_path = "/content/archive.rar"
extract_path = "/content/archive"

print(f"Attempting to download file with ID: {file_id}")
gdown.download(
    id=file_id,
    output=rar_path,
    quiet=False
)

if not os.path.exists(rar_path):
    print(f"Error: Download failed. No file found at {rar_path}")
else:
    print(f"Downloaded file exists. Checking file type: {rar_path}")
    try:
        # Verify file type before attempting extraction
        file_type_output = subprocess.check_output(['file', rar_path], stderr=subprocess.STDOUT).decode().strip()
        print(f"File type: {file_type_output}")

        if "RAR archive data" not in file_type_output:
            print("Error: The downloaded file is not a RAR archive according to 'file' command.")
            print("Please double-check the file_id and ensure it points to a shareable RAR file.")
            os.remove(rar_path)
        else:
            # Create extraction directory if it doesn't exist
            os.makedirs(extract_path, exist_ok=True)
            print(f"Extracting RAR file to {extract_path}...")
            try:
                # Use 'unrar x' for extraction, which handles directories automatically
                subprocess.run(['unrar', 'x', rar_path, extract_path], check=True)
                print("Dataset downloaded and extracted successfully.")
            except subprocess.CalledProcessError as e:
                print(f"Error during RAR extraction: {e}")
                print("Please ensure the RAR file is not corrupted and the unrar utility is correctly installed.")
            except Exception as e:
                print(f"An unexpected error occurred during extraction: {e}")
    except Exception as e:
        print(f"Could not determine file type or an error occurred: {e}")
```
